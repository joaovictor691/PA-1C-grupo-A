```python
# crie uma lista vazia chamada frutas.
#utilizanado append(), adicione 5 frutas informada pelo o usuario
# ao final exiba a lista completa
```

```python

frutas = []

print("Por favor, digite 5 frutas:")
for i in range(5):
    while True:
        fruta = input(f"Digite a fruta {i+1}: ").strip()
        if fruta and not fruta.isdigit(): # Verifica se não está vazio e não é apenas números
            frutas.append(fruta)

print(f"Lista completa de frutas: {frutas}")
```

```python
produtos = []

print("Por favor, cadastre 5 produtos:")
for i in range(5):
    produto = input(f"Digite o produto {i+1}: ")
    produtos.append(produto)

print("\nProdutos cadastrados:")
for i, produto in enumerate(produtos):
    print(f"{i+1} - {produto}")
```

```python
# crie uma lista vazia chamada numeros
# peça o usuario para informar 10 numeros
# depois crie duas listas vazias chamadas pares e impares
# pecorra a lista original e coloque cada numero na lista correspondente
```

```python
numeros = []
pares = []
impares = []

print("Por favor, informe 10 números:")
for i in range(10):
    while True:
        try:
            numero = int(input(f"Digite o número {i+1}: "))
            numeros.append(numero)
            break
        except ValueError:
            print("Entrada inválida. Por favor, digite um número inteiro.")

for numero in numeros:
    if numero % 2 == 0:
        pares.append(numero)
    else:
        impares.append(numero)

print(f"\nNúmeros informados: {numeros}")
print(f"Números pares: {pares}")
print(f"Números ímpares: {impares}")
```

```python
#  faça um pequeno sitemas para cadastrar as notas de 05 alunos.
# o programa deverá armazenar as notas em uma lista e depois mostrar:

# todas as notas
# maior nota
# media da turma
# quantidade de alunos com nota maior ou igual a 6
# quantidade alunos com nota menor que 6
```

```python
notas = []

print("Por favor, informe as notas de 5 alunos:")
for i in range(5):
    while True:
        try:
            nota = float(input(f"Digite a nota do aluno {i+1} (0 a 10): "))
            if 0 <= nota <= 10:
                notas.append(nota)
                break
            else:
                print("A nota deve estar entre 0 e 10.")
        except ValueError:
            print("Entrada inválida. Por favor, digite um número.")

# Mostrar todas as notas
print(f"\nTodas as notas: {notas}")

# Maior nota
maior_nota = max(notas)
print(f"Maior nota: {maior_nota:.2f}")

# Média da turma
media_turma = sum(notas) / len(notas)
print(f"Média da turma: {media_turma:.2f}")

# Quantidade de alunos com nota maior ou igual a 6
alunos_aprovados = 0
for nota in notas:
    if nota >= 6:
        alunos_aprovados += 1
print(f"Alunos com nota >= 6: {alunos_aprovados}")

# Quantidade de alunos com nota menor que 6
alunos_reprovados = len(notas) - alunos_aprovados
print(f"Alunos com nota < 6: {alunos_reprovados}")
```
