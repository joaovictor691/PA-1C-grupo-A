```python
#escrever um programa que o usario digite um numero e o programa imprime de 0 ate o numero
numero = int(input("Digite um número: "))
for i in range(numero + 1):
    print(i)
```

```python
#escrever um programa que o usario digite um numero e o programa imprime em ordens decrescente ate 0
numero = int(input("Digite um número: "))
for i in range(numero, -1, -1):
    print(i)
```

```python
#faça um numero e mostre todos numeros pares ate o numero
numero = int(input("Digite um número: "))
for i in range(numero + 1):
    if i % 2 == 0:
        print(i)
```

```python
#faça um numero e calcule a sima de 1 ate o numero
numero = int(input("Digite um número: "))
soma = 0
for i in range(1, numero + 1):
    soma += i
print(f"A soma de 1 até {numero} é: {soma}")
```

```python
#faça uma palavra e conte as letras ex python a palavra tem 6 letras
palavra = input("Digite uma palavra: ")
quantidade_letras = len(palavra)
print(f"A palavra '{palavra}' tem {quantidade_letras} letras.")
```

```python
palavra = input("Digite uma palavra: ")
vogais = "aeiouAEIOU"
contador_vogais = 0
for letra in palavra:
    if letra in vogais:
        contador_vogais += 1
print(f"A palavra '{palavra}' tem {contador_vogais} vogais.")
```

```python
numeros = []
print("Por favor, digite 5 números:")
for i in range(5):
    while True:
        try:
            numero = float(input(f"Digite o {i+1}º número: "))
            numeros.append(numero)
            break
        except ValueError:
            print("Entrada inválida. Por favor, digite um número.")

print(f"Os números digitados foram: {numeros}")
```
