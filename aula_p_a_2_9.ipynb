```python
notas = [7.5, 8.0, 6.5, 9.0, 8.5]

soma = sum(notas)
media = soma / len(notas)
maior_nota = max(notas)
menor_nota = min(notas)

print(f"Soma: {soma}")
print(f"Média: {media:.2f}")
print(f"Maior nota: {maior_nota}")
print(f"Menor nota: {menor_nota}")
```

```python
lista_original = list(range(1, 21))

pares_for = []
for n in lista_original:
    if n % 2 == 0:
        pares_for.append(n)
print("Pares (for):", pares_for)

pares_comprehension = [n for n in lista_original if n % 2 == 0]
print("Pares (list comprehension):", pares_comprehension)
```

```python
numeros_7 = [-5, 10, -3, 8, 0, -1, 7, -9, 4]

positivos = [n for n in numeros_7 if n > 0]
negativos = [n for n in numeros_7 if n < 0]

print("Positivos:", positivos)
print("Negativos:", negativos)

zeros = [n for n in numeros_7 if n == 0]
print("Zeros:", zeros)
```

```python
numeros_8 = [15, 3, 89, 42, 7, 21, 64, 12, 5, 99]

print("Lista original:", numeros_8)

copia_original = numeros_8.copy()

crescente = sorted(copia_original)
decrescente = sorted(copia_original, reverse=True)

print("Crescente:", crescente)
print("Decrescente:", decrescente)
print("Maior valor:", max(copia_original))
print("Menor valor:", min(copia_original))
print("Lista original intacta:", numeros_8)
```

```python
print("\n--- SISTEMA DE NOTAS ---")
qtd_alunos = int(input("Quantos alunos serão cadastrados? "))

alunos_dados = []

for i in range(qtd_alunos):
    nome = input(f"Nome do aluno {i+1}: ")
    nota = float(input(f"Nota do aluno {i+1}: "))
    alunos_dados.append({"nome": nome, "nota": nota})

print("\n--- RELATÓRIO FINAL ---")
print("Alunos cadastrados:")
for a in alunos_dados:
    print(f"- {a['nome']}: {a['nota']}")

todas_notas = [a["nota"] for a in alunos_dados]
media_turma = sum(todas_notas) / len(todas_notas)

aluno_maior = max(alunos_dados, key=lambda x: x["nota"])
aluno_menor = min(alunos_dados, key=lambda x: x["nota"])

aprovados = [a["nome"] for a in alunos_dados if a["nota"] >= 7.0]
reprovados = [a["nome"] for a in alunos_dados if a["nota"] < 7.0]

print(f"\nMédia da turma: {media_turma:.2f}")
print(f"Maior nota: {aluno_maior['nome']} ({aluno_maior['nota']})")
print(f"Menor nota: {aluno_menor['nome']} ({aluno_menor['nota']})")
print(f"Aprovados (>= 7.0): {', '.join(aprovados) if aprovados else 'Nenhum'}")
print(f"Reprovados (< 7.0): {', '.join(reprovados) if reprovados else 'Nenhum'}")

print("\n--- PESQUISA ---")
busca = input("Digite o nome do aluno que deseja pesquisar: ").strip()
encontrado = False
```
