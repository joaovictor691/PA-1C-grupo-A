```python
# operadores arimeticos


# soma de dois numeros

# entrada
a = int (input ("digite o primeiro numero: "))
b = int (input ("digite o segundo numero: "))

#processamento

# soma de dois numeros

soma = a + b
SUBTRAÇAO = a - b
MULTIPLICAÇÃO = a * b
DIVISÃO = a / b
DIVISAO_INTEIRA = a // b
resto_da_divisao = a % b

# saida
print ("")
print ("---------------SOMA -------------")
print (f"a soma de {a} mais {b} é igual a: {soma}")
print ("---------------SUBTRAÇÃO -------------")
print (f"a subtração de {a} menos {b} é igual a: {SUBTRAÇAO}")
print ("---------------MULTIPLICAÇÃO -------------")
print (f" a multiplicação de {a} vezes {b} é igual a: {MULTIPLICAÇÃO}")
print ("---------------DIVISÃO -------------")
print (f"a divisão de {a} dividido {b} é igual a: {DIVISÃO}")
print ("---------------DIVISÃO INTEIRA -------------")
print (f"a divisão inteira de {a} dividido {b} é igual a: {DIVISAO_INTEIRA}")
print ("---------------RESTO DA DIVISÃO -------------")
print (f"o resto da divisão de {a} dividido {b} é igual a: {resto_da_divisao}")
print ("---------------------------------------")
```

```python
def fazer_triagem():
    print("--- SIMULADOR DE TRIAGEM MÉDICA ---")
    print("Por favor, insira as informações do paciente:")

    nome = input("Nome do paciente: ")
    idade = int(input("Idade do paciente: "))
    sintoma_principal = input("Qual o sintoma principal? (Ex: dor no peito, febre, tosse, sangramento, dificuldade respiratória) ").lower()

    # Input validation for gravidade_dor to ensure it's a number between 1 and 10
    while True:
        try:
            gravidade_dor = int(input("Qual a intensidade da dor em uma escala de 1 a 10? (1=leve, 10=insuportável) "))
            if 1 <= gravidade_dor <= 10:
                break
            else:
                print("Por favor, insira um número entre 1 e 10.")
        except ValueError:
            print("Entrada inválida. Por favor, insira um número.")

    prioridade = "Verde (Não Urgente)"
    observacoes = []

    # Lógica de Triagem
    if "dor no peito" in sintoma_principal or "dificuldade respiratória" in sintoma_principal or "perda de consciência" in sintoma_principal or "sangramento intenso" in sintoma_principal:
        prioridade = "Vermelho (Emergência - Risco de Vida!)"
        observacoes.append("Sintomas indicam emergência médica grave.")
    elif "febre alta" in sintoma_principal and idade < 5 or gravidade_dor >= 8 or "fratura exposta" in sintoma_principal or "dor abdominal intensa" in sintoma_principal:
        prioridade = "Amarelo (Urgência - Atenção Prioritária)"
        observacoes.append("Sintomas requerem atenção médica rápida.")
    elif "vomito" in sintoma_principal or "diarreia" in sintoma_principal and idade < 2:
        prioridade = "Amarelo (Urgência - Atenção Prioritária)"
        observacoes.append("Vômito/Diarreia em criança pequena requer atenção.")
    else:
        prioridade = "Verde (Não Urgente - Pode aguardar)"
        observacoes.append("Sintomas geralmente não indicam risco imediato.")

    if gravidade_dor >= 7 and prioridade != "Vermelho (Emergência - Risco de Vida!)":
        prioridade = "Amarelo (Urgência - Atenção Prioritária)" if prioridade == "Verde (Não Urgente - Pode aguardar)" else prioridade
        observacoes.append(f"Nível de dor ({gravidade_dor}/10) é alto.")

    print("\n--- RESULTADO DA TRIAGEM ---")
    print(f"Paciente: {nome}")
    print(f"Idade: {idade} anos")
    print(f"Sintoma Principal: {sintoma_principal}")
    print(f"Nível de Dor: {gravidade_dor}/10")
    print(f"Prioridade: {prioridade}")
    if observacoes:
        print("Observações:")
        for obs in observacoes:
            print(f"  - {obs}")
    print("-----------------------------\n")

# Initialize game_running to True to start the loop
game_running = True

while game_running:
    fazer_triagem()
    another_patient = input("Deseja triar outro paciente? (sim/não): ").strip().lower()
    if another_patient not in ['sim', 's']:
        game_running = False
        print("Encerrando o simulador de triagem.")
```
